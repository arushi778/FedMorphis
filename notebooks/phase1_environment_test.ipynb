### Quick Visualization Sanity Check
Generate a synthetic fundus-like test pattern to verify array manipulation and plotting functionality.

```python
# Generate a dummy canvas
test_image = np.zeros((100, 100, 3), dtype=np.uint8)

# Draw sample test shapes using cv2
cv2.circle(test_image, (50, 50), 30, (0, 255, 0), -1)
cv2.line(test_image, (20, 20), (80, 80), (255, 255, 255), 2)

# Display using matplotlib
plt.figure(figsize=(3, 3))
plt.imshow(cv2.cvtColor(test_image, cv2.COLOR_BGR2RGB))
plt.title("OpenCV + Matplotlib Test")
plt.axis("off")
plt.show()

print("Sanity test completed successfully!")
```
